```python
#Exemplo
import pandas as pd
from sqlalchemy import create_engine

server = 'KETLEY'
database = 'AdventureWorksDW2019'

conexaoDB =  ('DRIVER={ODBC Driver 18 for SQL Server};'
                      f'SERVER=KETLEY;'
                      f'DATABASE=AdventureWorksDW2019;'
                      'Trusted_Connection=yes;' #conexão usuario do windows
                      #f'UID={username};'
                      #f'PWD={password};'                      
                      )

engine = create_engine(f'mssql+pyodbc:///?odbc_connect={conexaoDB}')
```

```python
conexaoDB =  ('DRIVER={ODBC Driver 17 for SQL Server};'f'SERVER=KETLEY;'f'DATABASE=AdventureWorksDW2019;''Trusted_Connection=yes;')

engine = create_engine(f'mssql+pyodbc:///?odbc_connect={conexaoDB}')

query = '''

SELECT * FROM [AdventureWorksDW2019].[dbo].[DimProduct]

'''

df = pd.read_sql_query(query, engine)

# Exiba  DataFrame
engine.dispose()
df.head()
```
