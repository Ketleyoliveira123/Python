```python

from selenium import webdriver
import chromedriver_autoinstaller

from selenium.webdriver.common.by import By 
from time import sleep # tempo de espera para carregar a página
from selenium.webdriver.common.keys import Keys #comandos do teclado

# Instalação automática do ChromeDriver (commicação entre cod e navegador)
chromedriver_autoinstaller.install()

# Inicializar o navegador Chrome
navegador = webdriver.Chrome()
```

```python
# Exemplo: abrindo paginas Web
Link = "https://rpachallenge.com/"
navegador.get(Link)
```

```python
sleep(5) # tempo de espera da pagina
```

```python
empresa = navegador.find_element(by= By.ID,value="Abiuv").send_keys("Teste")
empresa.clear() # esse comando vai limpar o campo empresa
```

```python
papelcompania = navegador.find_element(by= By.XPATH,value='//*[@id="eSJif"]') # esse comando vai localizar o campo papel da companhia
papelcompania.send_keys("Industrial") # esse comando vai preencher o campo papel da companhia com o valor "Industrial"

```

```python
sobrenome = navegador.find_element(by= By.XPATH,value='//*[@id="oQ9P5"]') # esse comando vai localizar o campo sobrenome
sobrenome.send_keys("Alves")
```

```python
Enviar = navegador.find_element(by= By.XPATH,value='/html/body/app-root/div[2]/app-rpa1/div/div[2]/form/input') # esse comando vai localizar o botão enviar
Enviar.click() 
```
