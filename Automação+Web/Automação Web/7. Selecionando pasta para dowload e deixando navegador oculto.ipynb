```python

from selenium import webdriver
from selenium.webdriver.chrome.options import Options as ChromeOptions
import chromedriver_autoinstaller
from time import sleep

# Instalar o ChromeDriver automaticamente
chromedriver_autoinstaller.install()

# Definir a pasta de destino
pasta_destino = r"C:\Users\Ketle\OneDrive\Documentos\Kety"

# Configurar as preferências de download
preferencias = {
    "download.default_directory": pasta_destino,
    "download.prompt_for_download": False,
    "download.directory_upgrade": True
}

# Configurar as opções do Chrome
options = ChromeOptions() # esse comando vai criar um objeto de opções do Chrome
options.add_experimental_option("prefs", preferencias) # esse comando vai adicionar as preferências de download ao objeto de opções do Chrome
options.headless = False # esse comando vai definir se o navegador será executado em modo headless (sem interface gráfica) ou não

# Inicializar o navegador COM as opções configuradas
navegador = webdriver.Chrome(options=options)

# Acessar o site
Link = "https://rpachallenge.com/"
navegador.get(Link)

```
